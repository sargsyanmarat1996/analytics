Что было сделано:

- Unpivot таблицы financial data, чтобы можно было использовать её в паре с таблицей prolongation. Нетронутыми остались все столбцы без значений.
- В оригинальные данные были внесены небольшие изменения: столбцы датой в формат даты, значения избавлены от текста.
- В таблицу financial data был добавлен столбец month (prolongation), которая заполняла месяцем окончания из таблицы prolongations в зависимости от id. Так как в задании не было указано про возобновление проектов, было выбрано первое окончание. Код: =ЕСЛИОШИБКА(ВПР([@id];prolongations!A:C;2;);" ")
- Туда же был добавлен столбец Значение окончания, который добавляет значение отгрузки в случае, если месяц отгрузки совпадает с месяцем окончания проекта. Код: =ЕСЛИ([@[Месяц окончания]]=[@Месяц];[@[Значения числа]];0)
- Туда же были добавлены столбцы с расчётом коэффицентов в случае продления в первый или второй месяц. Формулы проверяют совпадение месяца и месяца окончания, и id месяца окончания с id строчкой под ним. Далее к дате окончания прибавляем 1 или 2 месяца, и если под строчкой окончания оказывался необходимая нам дата, вытаскиываем числовое значение. Код: =ЕСЛИ([@[Месяц окончания]]=[@Месяц];[@[Значения числа]];0) =ЕСЛИ(И([@[Месяц окончания]]=[@Месяц];[@id]=A3;ДАТАМЕС([@Месяц];2)=D3);F3;0)
- В таблицах самих отчётов мы просто суммировали столбцы с фильтром под наши параметры. К примеру, чтобы месяц окончания совпадал с выбранным месяцем, как в коде: =ЕСЛИОШИБКА(СУММЕСЛИ('financial data unpivot'!G:G;A4;'financial data unpivot'!I:I);0)
- Или совпадали данные АМ, в годовом отчёте для менеджера =ЕСЛИОШИБКА(СУММЕСЛИ('financial data unpivot'!C:C;'Менеджер за год'!A3;'financial data unpivot'!I:I);0)
- В финальном отчёте по месяцам для каждого менеджера мы делали всё тоже самое, но высчитывали коэффицент сразу в одной ячейке. То есть сначала высчитывали отгрузку по окончания на месяц, и делили на неё отгрузку месяца или двух до неё. Код: =ЕСЛИОШИБКА(СУММЕСЛИМН('financial data unpivot'!$J:$J;'financial data unpivot'!$C:$C;'Менеджеры по месяцам'!$A9;'financial data unpivot'!$G:$G;'Менеджеры по месяцам'!C$3)/СУММЕСЛИМН('financial data unpivot'!$I:$I;'financial data unpivot'!$C:$C;'Менеджеры по месяцам'!$A9;'financial data unpivot'!$G:$G;'Менеджеры по месяцам'!C$3);0)
- Были построенный графики для таблицы Весь отдел и Менеджеры за год. По ним можно отследить: плавный прирост коэффицентра продления к марту и ноябрю месяцу, и что большинство менеджеров держатся ближе к показателю 0.5 коэффицентра продления, однако имеется и выброс, в виде коэффциента > 100%.
